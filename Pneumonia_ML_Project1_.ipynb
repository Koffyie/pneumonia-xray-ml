```python
import os

#cell 8

# Assuming the structure is /content/drive/MyDrive/Pneumonia-ML-Project1-/data/chest_xray/
base_dir = '/content/drive/MyDrive/Pneumonia-ML-Project1-/data/chest_xray'
train_dir = os.path.join(base_dir, 'train')
val_dir = os.path.join(base_dir, 'val')
test_dir = os.path.join(base_dir, 'test')

# Adding a diagnostic print to verify the new base_dir content
print(f"Checking contents of: {base_dir}")
if os.path.exists(base_dir):
    print("Contents:", os.listdir(base_dir))
else:
    print(f"Error: The directory {base_dir} does not exist. Please check your unzipping process or path.")


# Original code with corrected paths
if os.path.exists(train_dir):
    print("Train classes:", os.listdir(train_dir))
    print("Number of train images (NORMAL):", len(os.listdir(os.path.join(train_dir, 'NORMAL'))))
    print("Number of train images (PNEUMONIA):", len(os.listdir(os.path.join(train_dir, 'PNEUMONIA'))))
else:
    print(f"Error: Train directory not found at {train_dir}.")

if not os.path.exists(val_dir):
    print(f"Error: Validation directory not found at {val_dir}.")
if not os.path.exists(test_dir):
    print(f"Error: Test directory not found at {test_dir}.")
```

```python
import tensorflow as tf

#cell 9

IMG_SIZE = (150, 150)
BATCH_SIZE = 32

train_data = tf.keras.utils.image_dataset_from_directory(
    train_dir, image_size=IMG_SIZE, batch_size=BATCH_SIZE, label_mode='binary'
)
val_data = tf.keras.utils.image_dataset_from_directory(
    val_dir, image_size=IMG_SIZE, batch_size=BATCH_SIZE, label_mode='binary'
)
test_data = tf.keras.utils.image_dataset_from_directory(
    test_dir, image_size=IMG_SIZE, batch_size=BATCH_SIZE, label_mode='binary'
)
```

```python
#cell 10
for images, labels in train_data.take(1):
    print("Image shape:", images.shape)
    print("Label shape:", labels.shape)
```

```python
normalization_layer = tf.keras.layers.Rescaling(1./255)
#cell 11
train_data = train_data.map(lambda x, y: (normalization_layer(x), y))
val_data = val_data.map(lambda x, y: (normalization_layer(x), y))
test_data = test_data.map(lambda x, y: (normalization_layer(x), y))

print("Normalization applied")
```

```python
# LOGISTIC REGRESSION
# Converting the datasets to Numpy

#cell 12

import numpy as np

def dataset_to_numpy(dataset):
    images = []
    labels = []

    for batch_images, batch_labels in dataset:
        images.append(batch_images.numpy())
        labels.append(batch_labels.numpy())

    images = np.concatenate(images, axis=0)
    labels = np.concatenate(labels, axis=0).flatten()

    return images, labels
```

```python
X_train_img, y_train = dataset_to_numpy(train_data)
X_test_img, y_test = dataset_to_numpy(test_data)

#cell 13

# Re-split train into proper train/val (original val folder only had 16 images)
from sklearn.model_selection import train_test_split

X_train_img, X_val_img, y_train, y_val = train_test_split(
    X_train_img, y_train,
    test_size=0.15,
    stratify=y_train,
    random_state=42
)

print("Training:", X_train_img.shape, y_train.shape)
print("Validation:", X_val_img.shape, y_val.shape)
print("Testing:", X_test_img.shape, y_test.shape)
```

```python
# Normalizing the images

#cell 14

X_train_img = X_train_img
X_val_img = X_val_img
X_test_img = X_test_img
```

```python
# Flattening the images

#cell 15

X_train = X_train_img.reshape(X_train_img.shape[0], -1)
X_val = X_val_img.reshape(X_val_img.shape[0], -1)
X_test = X_test_img.reshape(X_test_img.shape[0], -1)

print("X_train:", X_train.shape)
print("X_val:", X_val.shape)
print("X_test:", X_test.shape)
```

```python
# Training
#cell 16
from sklearn.linear_model import LogisticRegression

logistic_model = LogisticRegression(
    max_iter=1000,
    class_weight='balanced',
    solver='liblinear'
)

print("Training Logistic Regression...")
logistic_model.fit(X_train, y_train)
print("Training complete!")

# Predicting

y_pred = logistic_model.predict(X_test)
```

```python
# Evaluating
#cell 17
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print("Logistic Regression Results")
print("---------------------------")
print(f"Accuracy:  {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall:    {recall:.4f}")
print(f"F1-score:  {f1:.4f}")

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    target_names=["NORMAL", "PNEUMONIA"]
))
```

```python
# Confusion Matrix
#cell 18
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, y_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["NORMAL", "PNEUMONIA"]
)

disp.plot()
plt.title("Logistic Regression Confusion Matrix")
plt.show()
```

**Transfer learning section using the ResNet50, for medical images.**

```python
#cell 29

import tensorflow as tf
from tensorflow.keras.applications import ResNet50
from tensorflow.keras import layers, models

# 1. Load Pre-trained Base Model (ResNet50 trained on ImageNet for x-rays)
base_model = ResNet50(
    weights='imagenet',
    include_top=False,
    input_shape=(150, 150, 3)
)

# 2. Freeze base model weights so they are not updated during initial training
base_model.trainable = False

# 3. Build Custom Classifier Head on top of base model
inputs = tf.keras.Input(shape=(150, 150, 3))
x = base_model(inputs, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.3)(x)
outputs = layers.Dense(1, activation='sigmoid')(x)

model = models.Model(inputs, outputs)

# 4. Compile Model
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-5),
    loss='binary_crossentropy',
    metrics=['accuracy', tf.keras.metrics.Precision(), tf.keras.metrics.Recall()]
)

model.summary()

# 5. Train the Classification Head
history = model.fit(
    train_data,
    epochs=3,
    validation_data=val_data
)

# 6. Fine-Tuning Step (Unfreeze Base Model for better accuracy)
base_model.trainable = True

# Recompile with a smaller learning rate to preserve learned features
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-5),
    loss='binary_crossentropy',
    metrics=['accuracy', tf.keras.metrics.Precision(), tf.keras.metrics.Recall()]
)
"""
fine_tune_history = model.fit(
    train_data,
    epochs=5,
    validation_data=val_data
)
"""
# 7. Evaluate on Test Dataset
test_results = model.evaluate(test_data)
print(f"\nTest Loss: {test_results[0]:.4f}")
print(f"Test Accuracy: {test_results[1]:.4f}")
print(f"Test Precision: {test_results[2]:.4f}")
print(f"Test Recall: {test_results[3]:.4f}")
```

EVALUATE TRANSFER LEARNING MODEL: Extract ground truth and model predictions on test_data

```python
#cell 30

import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report, roc_curve, auc, confusion_matrix

y_test_dl = []
y_pred_prob_dl = []

for images, labels in test_data:
    preds = model.predict(images, verbose=0)
    y_test_dl.extend(labels.numpy().flatten())
    y_pred_prob_dl.extend(preds.flatten())

y_test_dl = np.array(y_test_dl)
y_pred_prob_dl = np.array(y_pred_prob_dl)
y_pred_dl = (y_pred_prob_dl > 0.5).astype(int)

# Transfer Learning Metrics
acc_dl = accuracy_score(y_test_dl, y_pred_dl)
prec_dl = precision_score(y_test_dl, y_pred_dl)
rec_dl = recall_score(y_test_dl, y_pred_dl)
f1_dl = f1_score(y_test_dl, y_pred_dl)

# Get prediction probabilities for Classical Models (for ROC curves)
y_prob_log = logistic_model.predict_proba(X_test)[:, 1]
y_prob_rf = random_forest_model.predict_proba(X_test)[:, 1]
y_prob_svm = svm_pca_model.decision_function(X_test_pca)
```

```python
#cell 31

X_train_img_fixed = X_train_img
X_val_img_fixed = X_val_img
X_test_img_fixed = X_test_img

print('Scaling range for CNN (should be 0.0 to 1.0):')
print(X_train_img_fixed.min(), X_train_img_fixed.max())
print(X_val_img_fixed.min(), X_val_img_fixed.max())
```

```python
#cell 32

import tensorflow as tf
from tensorflow.keras import layers, models

# Build a CNN from scratch
cnn_model = models.Sequential([
    layers.Input(shape=X_train_img.shape[1:]),
    layers.Conv2D(32, (3, 3), activation='relu', padding='same'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu', padding='same'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(128, (3, 3), activation='relu', padding='same'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.5),
    layers.Dense(1, activation='sigmoid')
])

cnn_model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-4),
    loss='binary_crossentropy',
    metrics=['accuracy', tf.keras.metrics.AUC(name='auc')]
)
```

```python
from sklearn.utils.class_weight import compute_class_weight
import numpy as np

#cell 33

weights = compute_class_weight('balanced', classes=np.unique(y_train), y=y_train)
class_weight_dict = dict(enumerate(weights))
print(class_weight_dict)
```

```python
#cell 34

print("Class weights:")
print(class_weight_dict)

print("\nTraining class distribution:")
print(np.unique(y_train, return_counts=True))

print("\nValidation class distribution:")
print(np.unique(y_val, return_counts=True))

print("\nTest class distribution:")
print(np.unique(y_test, return_counts=True))
```

```python
#cell 35

early_stop = tf.keras.callbacks.EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True
)

history = cnn_model.fit(
    X_train_img_fixed, y_train,
    validation_data=(X_val_img_fixed, y_val),
    epochs=20,
    batch_size=32,
    class_weight=class_weight_dict,
    callbacks=[early_stop]
)
```

```python
import matplotlib.pyplot as plt

#cell 36

plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Val Loss')
plt.legend()
plt.title('Loss over epochs')
plt.show()

plt.plot(history.history['accuracy'], label='Train Accuracy')
plt.plot(history.history['val_accuracy'], label='Val Accuracy')
plt.legend()
plt.title('Accuracy over epochs')
plt.show()
```
